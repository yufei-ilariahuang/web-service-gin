```python
#TEST1: small txt file : example.txt(163kb)
# -----------------------------
# CELL 1: SETUP AND IMPORTS
# -----------------------------
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Ensure plots are displayed inline in the notebook
%matplotlib inline 

print("Libraries imported successfully. Time to run the analysis!")
```

```python
# -----------------------------
# CELL 2: LOAD AND PREPARE DATA
# -----------------------------

def load_time(filename):
    """Safely loads time data from a file."""
    try:
        with open(filename, 'r') as f:
            return float(f.read().strip())
    except FileNotFoundError:
        print(f"Error: {filename} not found. Ensure the setup scripts have been run.")
        return 0.0
    except ValueError:
        print(f"Error: {filename} contains invalid data.")
        return 0.0

# Load the times
local_time = load_time('local_time.txt')
distributed_time = load_time('distributed_time.txt')

if local_time == 0.0 or distributed_time == 0.0:
    print("\n[Data Load Failed] Cannot plot. Check your files and ensure both have been generated.")
else:
    # Create a DataFrame for easy plotting
    labels = ['Single-threaded Local', 'Distributed ECS (3-way Parallel)']
    times = [local_time, distributed_time]
    
    df_results = pd.DataFrame({
        'Execution Method': labels,
        'Time (seconds)': times
    })
    
    print("\nData loaded successfully:")
    print(df_results)
```

```python
# -----------------------------
# CELL 3: VISUALIZE THE RESULTS
# -----------------------------

if 'df_results' in locals() and not df_results.empty and df_results['Time (seconds)'].sum() > 0:
    
    # Configuration
    colors = ['#1f77b4', '#ff7f0e'] # Blue and Orange
    max_time = df_results['Time (seconds)'].max()
    
    plt.figure(figsize=(10, 6))
    
    # Create the bars
    bars = plt.bar(
        df_results['Execution Method'], 
        df_results['Time (seconds)'], 
        color=colors, 
        width=0.5
    )
    
    # Add labels and title
    plt.xlabel('Execution Method', fontsize=14)
    plt.ylabel('Time (seconds)', fontsize=14)
    plt.title('Performance Speedup: Local vs. Parallel Distributed Word Count', fontsize=16)
    
    # Add grid lines for readability
    plt.grid(axis='y', linestyle='--', alpha=0.6)
    
    # Add the time value on top of each bar
    for bar in bars:
        yval = bar.get_height()
        # Position the text slightly above the bar
        plt.text(bar.get_x() + bar.get_width()/2, yval + (max_time * 0.02), 
                 f'{yval:.3f}s', 
                 ha='center', va='bottom', fontsize=12, fontweight='bold')
    
    # Calculate and display speedup as a key metric
    speedup = local_time / distributed_time
    plt.text(0.5, max_time * 0.95, 
             f'Speedup Factor: {speedup:.2f}x', 
             ha='center', va='top', fontsize=14, color='green', 
             bbox=dict(facecolor='white', alpha=0.8, boxstyle='round,pad=0.5'))
    
    # Set the y-axis limit
    plt.ylim(0, max_time * 1.25)
    
    plt.show()

else:
    print("Cannot plot: Time data is missing or invalid. Rerun Cell 2.")
```
