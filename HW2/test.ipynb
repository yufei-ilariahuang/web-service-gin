```python
import requests

# URL for your local server
EC2_URL1 = "http://localhost:8080/albums"

# URL for your AWS server (replace with your actual public DNS or IP)
EC2_URL2 = "http://ec2-44-247-95-89.us-west-2.compute.amazonaws.com:8080/albums"

POST_DATA = {
    "id": "4",
    "title": "The Modern Sound of Betty Carter",
    "artist": "Betty Carter",
    "price": 49.99
}
HEADERS = {
    "Content-Type": "application/json"
}


def data_from_both(url1, url2):
    print("--- Checking Local Server ---")
    try:
        response = requests.get(url1)
        print(f"Localhost response: {response.text}")
    except requests.exceptions.RequestException as e:
        print(f"Error occurred while testing {url1}: {e}")

    print("\n--- Checking AWS Server ---")
    try:
        response = requests.get(url2)
        print(f"AWS EC2 response: {response.text}")
    except requests.exceptions.RequestException as e:
        print(f"Error occurred while testing {url2}: {e}")
    return

```

```python
# -------------------------------------------------------------
# PART 2: EXECUTION (This is the part you are likely missing!)
# -------------------------------------------------------------

print("============== STARTING DATA TEST ==============")
# This line calls the function to do the first check
data_from_both(EC2_URL1, EC2_URL2)

print("\n\n============== ADDING DATA TO AWS SERVER ONLY ==============")
try:
    # This line sends the POST request
    response = requests.post(EC2_URL2, json=POST_DATA, headers=HEADERS, timeout=5)
    print(f"POST request to AWS response: {response.text}")
except requests.exceptions.RequestException as e:
    print(f"Error occurred while posting to {EC2_URL2}: {e}")

print("\n\n============== CHECKING DATA AGAIN ==============")
# This line calls the function again to show the difference
data_from_both(EC2_URL1, EC2_URL2)

print("\n\nUh oh... what happened?")
```
