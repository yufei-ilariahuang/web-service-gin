```python
# -----------------------------
# CELL 1: SETUP AND IMPORTS
# -----------------------------
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Ensure plots are displayed inline in the notebook
%matplotlib inline 

print("Libraries imported successfully. Time to run the analysis!")
```

```python
# -----------------------------
# CELL 2: LOAD AND PREPARE DATA
# -----------------------------

def load_time(filename):
    """Safely loads time data from a file."""
    try:
        with open(filename, 'r') as f:
            return float(f.read().strip())
    except FileNotFoundError:
        print(f"Error: {filename} not found. Ensure the setup scripts have been run.")
        return 0.0
    except ValueError:
        print(f"Error: {filename} contains invalid data.")
        return 0.0

# Load the times
local_time = load_time('local_time.txt')
distributed_time = load_time('distributed_time.txt')

if local_time == 0.0 or distributed_time == 0.0:
    print("\n[Data Load Failed] Cannot plot. Check your files and ensure both have been generated.")
else:
    # Create a DataFrame for easy plotting
    labels = ['Single-threaded Local', 'Distributed ECS (3-way Parallel)']
    times = [local_time, distributed_time]
    
    df_results = pd.DataFrame({
        'Execution Method': labels,
        'Time (seconds)': times
    })
    
    print("\nData loaded successfully:")
    print(df_results)
```
